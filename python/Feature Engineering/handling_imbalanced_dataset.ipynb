### Handling Imbalanced Dataset

1. Up Sampling ( increasing data point of minority)

2.Down sampling (decreasing data point of majority)

```python
import numpy as np
import pandas as pd

## set the random seed for reproducibility
np.random.seed(123)

## create a sample dataframe with two classes
n_samples = 1000
class_0_ratio=0.9
n_class_0 = int(n_samples * class_0_ratio)
n_class_1 = n_samples - n_class_0
```

```python
n_class_0, n_class_1
```

```python
## create my dataframe with imbalanced classes
class_0 = pd.DataFrame({
    'feature1': np.random.normal(loc=0, scale=1, size=n_class_0),
    'feature2': np.random.normal(loc=0, scale=1, size=n_class_0),
    'target': [0] * n_class_0
})
class_1 = pd.DataFrame({
    'feature1': np.random.normal(loc=1, scale=1, size=n_class_1),
    'feature2': np.random.normal(loc=1, scale=1, size=n_class_1),
    'target': [1] * n_class_1
})
```

```python
df=pd.concat([class_0, class_1]).reset_index(drop=True)
```

```python
df.head()
```

```python
df.tail()
```

```python
df['target'].value_counts()
```

## upsampling


```python
df_minority = df[df['target'] == 1]
df_majority = df[df['target'] == 0]
```

```python
from sklearn.utils import resample

df_minority_upsample=df_minority_upsampled = resample(df_minority, replace=True, n_samples=len(df_majority), random_state=42)
```

```python
df_minority_upsample.shape
```

```python
df_minority_upsample.head()
```

```python
df_upsampled=pd.concat([df_majority, df_minority_upsample])
```

```python
df_upsampled
```

```python
df_upsampled['target'].value_counts()
```
