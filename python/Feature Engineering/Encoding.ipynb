## Nominal / OHE Encoding

It is a technique used to represent categorical data as numerical data, which is more suitable for ML algorithms. In this technique each category is represented as binary vector where each bit corresponds to a unique category. For example if we have categorical variable 'color' with three possible values (red, green,blue) we can represent it in ohe as follows:
1. Red (1,0,0)

2. Green (0,1,0)

3. Blue (0,0,1)


```python
import pandas as pd
from sklearn.preprocessing import OneHotEncoder
```

```python
## create a sample dataframe
df = pd.DataFrame({'color': ['red', 'blue', 'green','green', 'red', 'blue']})
```

```python
df.head()
```

```python
## create an instance of OneHotEncoder
encoder = OneHotEncoder()
```

```python
## perform fit and transform
encoded_data = encoder.fit_transform(df[['color']]).toarray()
encoded_data
```

```python
import pandas as pd
encoder_df=pd.DataFrame(encoded_data,columns=encoder.get_feature_names_out(['color']))
encoder_df
```

```python
## for new data
encoder.transform([['red']]).toarray()
```

```python
pd.concat([df, encoder_df], axis=1)
```

### Label Encoding

Label encoding is a technique used to convert categorical data into numerical data by assigning a unique integer to each category. For example if we have a categorical variable 'color' with three possible values (red, green, blue) we can represent it in label encoding as follows:
1. Red (0)
2. Green (1)
3. Blue (2)

```python
df.head()
```

```python
from sklearn.preprocessing import LabelEncoder
```

```python
label_encoder = LabelEncoder()
color_encoded= label_encoder.fit_transform(df[['color']])
color_encoded
```

```python
label_encoder.transform([['red']])
```

```python
label_encoder.transform([['blue']])
```

## Ordinal Encoding

Ordinal encoding is a technique used to convert categorical data into numerical data by assigning a unique integer to each category based on their order or rank. For example if we have a categorical variable 'size' with three possible values (small, medium, large) we can represent it in ordinal encoding as follows:
1. Small (0)
2. Medium (1)
3. Large (2)

```python
from sklearn.preprocessing import OrdinalEncoder
```

```python
df=pd.DataFrame({'size': ['small', 'medium', 'large','medium', 'small', 'large']})
```

```python
df
```

```python
ordinal_encoder = OrdinalEncoder(categories=[['small', 'medium', 'large']])
size_encoded = ordinal_encoder.fit_transform(df[['size']])
size_encoded
```

```python
ordinal_encoder.transform([['medium']])
```

## Target Guided Ordinal Encoding
Target guided ordinal encoding is a technique used to convert categorical data into numerical data by assigning a unique integer to each category based on the mean of the target variable for that category. For example if we have a categorical variable 'color' with three possible values (red, green, blue) and a target variable 'price' we can represent it in target guided ordinal encoding as follows:
1. Red (mean price for red)
2. Green (mean price for green)
3. Blue (mean price for blue)

```python
import pandas as pd

df=pd.DataFrame({
    'city': ['New York', 'London', 'Paris','Tokyo', 'New York', 'Paris'],
'price':[100, 200, 150, 120, 180, 130]})
```

```python
df
```

```python
mean_prices = df.groupby('city')['price'].mean().to_dict()
```

```python
mean_prices
```

```python
df['city_encoded'] = df['city'].map(mean_prices)
```

```python
df
```

```python
df[['price', 'city_encoded']]
```
