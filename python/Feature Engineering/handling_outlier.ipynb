## Minimum, Maximum, Median, Q1, Q2, IQR

```python
import numpy as np
```

```python
lst_marks=[45,32,56,75,89,90,87,67,54,45,98,99,67,74]
minimum,maximum,Q1,Q3,median = np.quantile(lst_marks,[0,1,0.25,0.75,0.5])
```

```python
minimum,maximum,Q1,Q3,median
```

```python
IQR = Q3-Q1
print("IQR:",IQR)
lower_bound = Q1 - 1.5*IQR
print("Lower Bound:",lower_bound)
upper_bound = Q3 + 1.5*IQR
print("Upper Bound:",upper_bound)
```

```python
import seaborn as sns
sns.boxplot(lst_marks)
```
