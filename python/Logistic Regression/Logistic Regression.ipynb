## Logistic Regression Implementation

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
%matplotlib inline
```

```python
from sklearn.datasets import make_classification
```

```python
## create dataset
X, y = make_classification(n_samples=1000, n_features=10, n_classes=2, random_state=42)
```

```python
pd.DataFrame(X)
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.30, random_state=42)
```

```python
## model training
from sklearn.linear_model import LogisticRegression
logistic=LogisticRegression()
logistic.fit(X_train, y_train)
```

```python
y_pred=logistic.predict(X_test)
print("Predicted values are:",y_pred)
```

```python
logistic.predict_proba(X_test)
```

```python
from sklearn.metrics import classification_report, confusion_matrix,accuracy_score
```

```python
score=accuracy_score(y_test,y_pred)
print("Accuracy score is:",score)
cm=confusion_matrix(y_test,y_pred)
print("Confusion matrix is:",cm)
print("Classification report is:",classification_report(y_test,y_pred)) 
```

### Hyper Parameter Tuning & Cross Validation

```python
model=LogisticRegression()
penalties=['l1','l2','elasticnet']
c_values=[100,10,1.0,0.1,0.01]
solvers=['newton-cg','lbfgs','liblinear','sag','saga']
```

```python
param_grid = dict(penalty=penalties, C=c_values, solver=solvers)
param_grid
```

```python
from sklearn.model_selection import StratifiedKFold
cv=StratifiedKFold()
```

```python
from sklearn.model_selection import GridSearchCV
grid=GridSearchCV(estimator=model,param_grid=param_grid,cv=cv,scoring='accuracy',n_jobs=-1)
```

```python
grid
```

```python
grid.fit(X_train,y_train)
```

```python
grid.best_params_
```

```python
y_pred=grid.predict(X_test)
```

```python
score=accuracy_score(y_test,y_pred)
print("Accuracy score is:",score)
cm=confusion_matrix(y_test,y_pred)
print("Confusion matrix is:",cm)
print("Classification report is:",classification_report(y_test,y_pred)) 
```

## Randomised Search CV

## Logistic Regression For Multiclass Classification

```python
X, y = make_classification(n_samples=1000, n_features=10, n_informative=3, n_classes=3, random_state=15)
```

```python
X
```

```python
y
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.30, random_state=42)
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.multiclass import OneVsRestClassifier
logistic = OneVsRestClassifier(LogisticRegression())
logistic.fit(X_train, y_train)
y_pred=logistic.predict(X_test)
```

```python
y_pred
```

```python
score=accuracy_score(y_test,y_pred)
print("Accuracy score is:",score)
cm=confusion_matrix(y_test,y_pred)
print("Confusion matrix is:",cm)
print("Classification report is:",classification_report(y_test,y_pred)) 
```

## Logistic Regression For Imbalanced Dataset

```python
from collections import Counter
from sklearn.datasets import make_classification
```

```python
X, y = make_classification(
    n_samples=10000,
    n_features=2,
    n_clusters_per_class=1,
    n_redundant=0,
    weights=[0.99],
    random_state=10
)
```

```python
X
```

```python
Counter(y)
```

```python
sns.scatterplot(
    x=pd.DataFrame(X)[0],
    y=pd.DataFrame(X)[1],
    hue=y
)
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.25, random_state=0)
```

```python
from sklearn.linear_model import LogisticRegression
model = LogisticRegression()
penalties = ['l1', 'l2', 'elasticnet']
c_values = [100, 10, 1.0, 0.1, 0.01]
solvers = ['newton-cg', 'lbfgs', 'liblinear', 'sag', 'saga']
class_weight=[{0:w,1:y} for w in [1,10,50,100] for y in [1,10,50,100]]
```

```python
param_grid = dict(penalty=penalties, C=c_values, solver=solvers, class_weight=class_weight)
```

```python
param_grid
```

```python
from sklearn.model_selection import GridSearchCV
from sklearn.model_selection import StratifiedKFold
cv=StratifiedKFold()
grid = GridSearchCV(estimator=model, param_grid=param_grid, cv=cv, scoring='accuracy', n_jobs=-1)
```

```python
grid.fit(X_train,y_train)
```

```python
grid.best_params_
```

```python
grid.best_score_
```

```python
y_pred=grid.predict(X_test)
```

```python
from sklearn.metrics import classification_report, confusion_matrix,accuracy_score
```

```python
score=accuracy_score(y_test,y_pred)
print("Accuracy score is:",score)
print("Confusion matrix is:",confusion_matrix(y_test,y_pred))
print("Classification report is:",classification_report(y_test,y_pred))
```
