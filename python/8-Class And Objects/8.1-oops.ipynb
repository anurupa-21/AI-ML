## Classes & Objects
In object-oriented programming (OOP), classes and objects are fundamental concepts that allow developers to create modular, reusable, and organized code.

```python
## A class is a blueprint for creating objects. It defines a set of attributes and methods that the created objects will have. In Python, you can define a class using the `class` keyword followed by the class name and a colon. Here's an example of a simple class definition:
class Car:
    pass
audi=Car()
bmw=Car()
print(audi)
print(bmw)
print(type(audi))
print(type(bmw))
```

```python
audi.windows=4
audi.color='red'
print(audi.windows)
print(audi.color)
tata=Car()
print(tata.windows) # it will give error because windows is not defined for tata object
```

```python
### instance variable and instance method
class Dog:
    ## Constructor
    def __init__(self,name,age):
        self.name=name
        self.age=age
## creating object of Dog class
dog1=Dog('Tommy',5)
print(dog1.name)
print(dog1.age)

dog2=Dog()
print(dog2.name) # it will give error because name is not defined for dog2 object
print(dog2.age) # it will give error because age is not defined for dog2 object

```

```python
## define class with instance method
class Dog:
    ## Constructor
    def __init__(self,name,age):
        self.name=name
        self.age=age
    ## instance method
    def bark(self):
        print(f'{self.name} is barking')
## creating object of Dog class
dog1=Dog('Tommy',5)
print(dog1.name)
print(dog1.age)
dog1.bark()
```

```python
## modeling a bank account using class and objects
class BankAccount:
    def __init__(self,account_number,account_holder,balance):
        self.account_number=account_number
        self.account_holder=account_holder
        self.balance=balance
    def deposit(self,amount):
        self.balance+=amount
        print(f'Amount {amount} deposited successfully. Current balance is {self.balance}')
    def withdraw(self,amount):
        if amount>self.balance:
            print('Insufficient balance')
        else:
            self.balance-=amount
            print(f'Amount {amount} withdrawn successfully. Current balance is {self.balance}')
account1=BankAccount('1234567890','John Doe',1000)
print(account1.account_number)
print(account1.account_holder)
print(account1.balance)
account1.deposit(500) 
account1.withdraw(200)
account1.withdraw(1500) # it will give error because balance is insufficient  
```
