## File Operation- Read And Write Files


File handling is a crucial part of any programming language. Python provides built-in functions and methods to read from and write to files, both text and binary. This lesson will cover the basics of file handling, including reading and writing text files and binary files.

```python
## read a whole file
with open('example.txt', 'r') as file:
    content = file.read()
    print(content)  # Output: Contents of example.txt
```

```python
## read a file line by line
with open('example.txt', 'r') as file:
    for line in file:
        print(line.strip())  # Output: Each line of example.txt without leading/trailing whitespace
```

```python
## writing a file (overwriting)
with open('example.txt', 'w') as file:
    file.write("This is a new line.\n")  # Output: example.txt now contains "This is a new line."
    file.write("This will overwrite the existing content.\n")  # Output: example.txt now contains "This will overwrite the existing content."   
    file.write("This is the final line.\n")  # Output: example.txt now contains "This is the final line."
    print("File has been written.")  # Output: File has been written.
```

```python
## write file with append mode
with open('example.txt', 'a') as file:
    file.write("This line will be appended to the existing content.\n")  # Output: example.txt now contains the new line appended to the existing content.
    print("Line has been appended.")  # Output: Line has been appended.
```

```python
## writing list of lines to a file
lines = ["First line.\n", "Second line.\n", "Third line.\n"]
with open('example.txt', 'a') as file:
    file.writelines(lines)  # Output: example.txt now contains the three lines from the list.
    print("Lines have been written to the file.")  # Output: Lines have been written to the file.   
```

```python
## binary file operations
data = b'\x00\x01\x02\x03\x04'
with open('example.bin', 'wb') as file:
    file.write(data)  # Output: example.bin now contains the binary data
```

```python
with open('example.bin', 'rb') as file:
    binary_content = file.read()
    print(binary_content)  # Output: b'\x00\x01\x02\x03\x04'
```

```python
## read the content from source file and write to destination file
with open('example.txt', 'r') as source_file:
    content = source_file.read()
with open('destination.txt', 'w') as dest_file:
  dest_file.write(content)  # Output: destination.txt now contains the same content as source.txt
print("Content has been copied from source.txt to destination.txt.")  # Output: Content has been copied from source.txt to destination.txt.
```

```python
## read a text file and count the number of lines,words, and characters
with open('example.txt', 'r') as file:
    content = file.read()
    lines = content.splitlines()
    words = content.split()
    characters = len(content)
    print(f"Number of lines: {len(lines)}")  # Output: Number of lines in example.txt
    print(f"Number of words: {len(words)}")  # Output: Number of words in example.txt
    print(f"Number of characters: {characters}")  # Output: Number of characters in example.txt
```

```python
## writing and then reading a file
with open('example.txt', 'w+') as file:
    file.write("Hello, World!\nThis is a test file.\n")  # Output: example.txt now contains the two lines of text.
    file.seek(0)  # Move the file pointer back to the beginning of the file
    content = file.read()   
    print(content)  # Output: Contents of example.txt after writing
```
