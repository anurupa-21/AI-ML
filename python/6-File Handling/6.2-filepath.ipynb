```python
import os
new_directory="package"
os.mkdir(new_directory)
print(f"Directory '{new_directory}' has been created.")  # Output: Directory 'package
```

```python
## listing files and directories
items = os.listdir('.')
print("Items in the current directory:")
for item in items:
    print(item)  # Output: List of files and directories in the current directory
```

```python
## joining paths
file_name = "example.txt"
directory = "package"
full_path = os.path.join(directory, file_name)
print(f"Full path to the file: {full_path}")  # Output: Full path to the file: package/example.txt
```

```python
dir_name="folder"
file_name="file.txt"
full_path = os.path.join(os.getcwd(), dir_name, file_name)
print(f"Full path to the file: {full_path}")  # Output: Full path to the file: /current/working/directory/folder/file.txt
```

```python
path = "example1.txt"
if os.path.exists(path):
    print(f"The file '{path}' exists.")  # Output: The file 'example1.txt' exists.
else:
    print(f"The file '{path}' does not exist.")  # Output: The file 'example1.txt' does not exist.
```

```python
path = "example.txt"
if os.path.isfile(path):
    print(f"The path '{path}' is a file.")  # Output: The path 'example.txt' is a file.
elif os.path.isdir(path):
    print(f"The path '{path}' is a directory.")  # Output: The path 'example.txt' is a directory.
else:
    print(f"The path '{path}' does not exist.")  # Output: The path 'example.txt' does not exist.
```

```python
## getting the absolute path of a file
relative_path = "example.txt"
absolute_path = os.path.abspath(relative_path)
print(f"Absolute path of '{relative_path}': {absolute_path}")  # Output: Absolute path of 'example.txt': /full/path/to/example.txt
```
