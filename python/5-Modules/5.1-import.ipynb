## Importing Modules in Python: Modules & Packages

```python
import math
math.sqrt(16)  # Output: 4.0
```

```python
from math import sqrt, pi
print(sqrt(25))  # Output: 5.0
print(pi)  # Output: 3.141592653589793
```

```python
import numpy as np
array = np.array([1, 2, 3, 4, 5])
print(array)  # Output: [1 2 3 4 5]
```

```python
from math import *
sqrt(9)  # Output: 3.0
print(pi)  # Output: 3.141592653589793
```

```python
from packages.maths import addition
addition(2, 3)  # Output: 5
```

```python
from package import maths
maths.addition(2, 3)  # Output: 5
```
