## Standard Library overview

```python
import array
arr=array.array('i',[1,2,3,4,5])
print(arr)  # Output: array('i', [1, 2, 3
```

```python
import math as m
print(m.sqrt(36))  # Output: 6.0
print(m.pi)  # Output: 3.141592653589793
```

```python
import random
print(random.randint(1, 10))  # Output: Random integer between 1 and 10
print(random.choice(['apple', 'banana', 'cherry']))  # Output: Randomly chosen fruit from the list
```

```python
### file and directory access

import os
print(os.getcwd())  # Output: Current working directory
print(os.listdir())  # Output: List of files and directories in the current directory       
os.mkdir('new_directory')  # Creates a new directory named 'new_directory'
# os.rename('new_directory', 'renamed_directory')  # Renames 'new_directory' to 'renamed_directory'
```

```python
## high level file operations
import shutil
shutil.copyfile('source.txt', 'destination.txt')  # Copies source_file.txt
```

```python
## data serialization
import json
data = {'name': 'Alice', 'age': 30, 'city': 'New York'}
json_string = json.dumps(data)  # Convert Python dictionary to JSON string
print(json_string)  # Output: {"name": "Alice", "age": 30, "city": "New York"}
print(type(json_string))  # Output: <class 'str'>
json_data = json.loads(json_string)  # Convert JSON string back to Python dictionary
print(json_data)  # Output: {'name': 'Alice', 'age': 30, 'city': 'New York'}
print(type(json_data))  # Output: <class 'dict'>
```
