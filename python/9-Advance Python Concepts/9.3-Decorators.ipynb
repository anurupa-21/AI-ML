### Decorators
Decorators are a powerful and flexible feature in Python that allows you to modify the behavior of a function or class method. They are commonly used to add functionality to functions or methods without modifying their actual code. This lesson covers the basics of decorators, including how to create and use them.

```python

### function copy
### closures
### decorators
```

```python

## function copy
def welcome():
    return "Welcome to the advanced python course"

welcome()
```

```python
wel=welcome
print(wel())
del welcome
print(wel())

```

```python
## closures
def main_welcome():
    msg="Welcome"
    def sub_welcome_method():
        print("Welcome to the advance python course")
        print(msg)
        print("please learn these concepts properly")
    return sub_welcome_method()

```

```python
main_welcome()
```

```python

## closures
def main_welcome(msg):
    
    def sub_welcome_method():
        print("Welcome to the advance python course")
        print(msg)
        print("please learn these concepts properly")
    return sub_welcome_method()

```

```python
main_welcome("welcome everyone")
```

```python
## closures
def main_welcome(func):
   
    def sub_welcome_method():
        print("Welcome to the advance python course")
        func("Welcome everyone to this course")
        print("please learn these concepts properly")
    return sub_welcome_method()

```

```python
main_welcome(print)
```

```python
## closures
def main_welcome(func,lst):
   
    def sub_welcome_method():
        print("Welcome to the advance python course")
        print(func(lst))
        print("please learn these concepts properly")
    return sub_welcome_method()

```

```python
main_welcome(len,[1,2,3,4,5])
```

```python
## Decorators

def main_welcome(func):
   
    def sub_welcome_method():
        print("Welcome to the advance python course")
        func()
        print("please learn these concepts properly")
    return sub_welcome_method()

```

```python
def course_introduction():
    print("this is course intro")
course_introduction()
```

```python
main_welcome(course_introduction)
```

```python
@main_welcome
def course_introduction():
    print("this is course intro")


```

```python
## Decorator

def my_decorator(func):
    def wrapper():
        print("Something is happening before the function is called.")
        func()
        print("Something is happening after the function is called.")
    return wrapper
```

```python
@my_decorator
def say_hello():
    print("Hello!")
```

```python
say_hello()
```

```python

## Decorators WWith arguments
def repeat(n):
    def decorator(func):
        def wrapper(*args, **kwargs):
            for _ in range(n):
                func(*args, **kwargs)
        return wrapper
    return decorator
```

```python

@repeat(3)
def say_hello():
    print("Hello")
```

```python

say_hello()
```

### Conclusion
Decorators are a powerful tool in Python for extending and modifying the behavior of functions and methods. They provide a clean and readable way to add functionality such as logging, timing, access control, and more without changing the original code. Understanding and using decorators effectively can significantly enhance your Python programming skills.
