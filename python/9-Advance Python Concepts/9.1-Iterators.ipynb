## Iterators

Iterators are a powerful tool in programming that allow you to traverse through a collection of items, such as lists, arrays, or other data structures. They provide a way to access each element in a sequence without exposing the underlying structure of the collection.

```python
my_list = [1, 2, 3, 4, 5,6]
for i in my_list:
    print(i)
```

```python
type(my_list)
print(type(my_list))
print(my_list)
```

```python
iterator = iter(my_list)
print(iterator)
print(type(iterator))
```

```python
## iterate through all the elements in the iterator
print(next(iterator))
print(next(iterator))
print(next(iterator))
print(next(iterator))
print(next(iterator))
print(next(iterator)) 
print(next(iterator)) # this will raise StopIteration error because there are no more elements to iterate through          
```

```python
iterator = iter(my_list)
try:
    while True:
        element = next(iterator)
        print(element)
except StopIteration:
    print("No more elements to iterate through.")
```
