## Multithreading

When to use Multi Threading?

I/O-bound tasks: Tasks that spend more time waiting for I/O operations (e.g., file operations, network requests).

 Concurrent execution: When you want to improve the throughput of your application by performing multiple operations concurrently.


```python
import threading
import time
def print_numbers():
    for i in range(5):
        print(f"Numbers:{i}")
def print_letters():
    for letter in "abcde":
        print(f"Letter:{letter}")
t=time.time()
print_numbers()
print_letters()
finished_time=time.time()-t;
print(finished_time)
```

```python
def print_numbers():
    for i in range(5):
        time.sleep(2)
        print(f"Number:{i}\n")

def print_letter():
    for letter in "abcde":
        time.sleep(2)
        print(f"Letter: {letter}\n")

##create 2 threads
t1=threading.Thread(target=print_numbers)
t2=threading.Thread(target=print_letter)

t=time.time()
## start the thread
t1.start()
t2.start()

### Wait for the threads to complete
t1.join()
t2.join()

finished_time=time.time()-t
print(finished_time)
```
