## Dictionaries

```python
## creating a dictionary
empty_dict={}
print(empty_dict)
print(type(empty_dict))
```

```python
empty_dict=dict()
print(empty_dict)
print(type(empty_dict))
```

```python
student={"name":"Alice","age":20,"grade":"A"}
print(student)
print(type(student))
```

```python
## Error: keys must be immutable
student={"name":"Alice","age":20,"name":"Bob"}
print(student)
```

```python
## accessing dictionary elements
student={"name":"Alice","age":20,"grade":"A"}
print(student["name"])
print(student["age"])
print(student["grade"])

```

```python
## accessing using get method
print(student.get("name"))
print(student.get("age"))
print(student.get("grade"))
print(student.get("address"))  # returns None if key does not exist
print(student.get("address","Not Found"))  # returns default value if key does not exist
```

```python
## modifying dictionary elements
# dictionaries are mutable, we can change their values
student={"name":"Alice","age":20,"grade":"A"}
print(student)
student["name"]="Bob"
student["age"]=21
student["grade"]="B"
print(student)
student["address"]="123 Main St"
print(student)
```

```python
## delete a key-value pair
del student["grade"]
print(student)
```

```python
## Dictionary methods
student={"name":"Alice","age":20,"grade":"A"}
print(student.keys())  # returns a view of keys
print(student.values())  # returns a view of values 
print(student.items())  # returns a view of key-value pairs 
```

shallow copy of a dictionary

```python

## problem: both variables point to the same dictionary, changes to one will affect the other
student={"name":"Alice","age":20,"grade":"A"}
student_copy=student
print(student)
print(student_copy)
student["name"]="Bob"
print(student)
print(student_copy)  # student_copy also changes because it is a reference to the same dictionary
```

```python
student={"name":"Alice","age":20,"grade":"A"}
print(student)  
student_copy=student.copy()  # creates a shallow copy of the dictionary
print(student_copy)
student["name"]="Bob"
print(student)  
print(student_copy)  # student_copy does not change because it is a separate copy of the dictionary
```

```python
### iterating over a dictionary
student={"name":"Alice","age":20,"grade":"A"}
print("Iterating using for loop:")
for key in student:
    print(key, student[key])
print("Iterating using items() method:")
for key, value in student.items():
    print(key, value)
print("Iterating using values() method:")
for value in student.values():
    print(value)
print("Iterating using keys() method:")
for key in student.keys():
    print(key)
```

```python
# nested dictionaries
students={
    "student1":{"name":"Alice","age":20,"grade":"A"},
    "student2":{"name":"Bob","age":22,"grade":"B"}
}   
print(students)
print(students["student1"]["name"])
print(students["student2"]["age"])
print(students["student1"]["grade"])
print(students["student2"]["name"])
print(students["student1"]["age"])
print(students["student2"]["grade"])
print(students["student1"]["name"], students["student1"]["age"], students["student1"]["grade"])
print(students["student2"]["name"], students["student2"]["age"], students["student2"]["grade"])
```

```python
## iterating over nested dictionaries
print("Iterating over nested dictionaries:")
for student_id, student_info in students.items():
    print(student_id, student_info)
    for key, value in student_info.items():
        print(f"  {key}: {value}")
```

```python
## dictionary comprehension
squares={x:x**2 for x in range(1,6)}
print(squares)
```

```python
even_squares={x:x**2 for x in range(10) if x%2==0}
print(even_squares)
```

```python
## practice problems
# 1. Create a dictionary to store the names and ages of three people. Then,
#    print the name and age of each person.
people={"Alice":20,"Bob":22,"Charlie":25}           
for name, age in people.items():
    print(f"Name: {name}, Age: {age}")
    
```

```python
##2. use a dictionary to count the frequency of each word in a given sentence. Then, print the word counts.
sentence="hello world hello python"
word_counts={}
for word in sentence.split():
    if word in word_counts:
        word_counts[word]+=1
    else:
        word_counts[word]=1
print(word_counts)
```

```python
## use a dictionary to count the frequency of elements in a list. Then, print the counts.
numbers=[1,2,2,3,3,3,4,4,4,4]
number_counts={}
for number in numbers:
    if number in number_counts:
        number_counts[number]+=1
    else:
        number_counts[number]=1
print(number_counts)
```

```python
## merge two dictionaries
dict1={"a":1,"b":2}
dict2={"b":3,"c":4}
merged_dict={**dict1, **dict2}  # dict2 values will overwrite dict1 values for duplicate keys
print(merged_dict)
```
