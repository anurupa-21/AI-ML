## Sets


```python
## create a set
my_set={1,2,3,4,5}
print(my_set)
print(type(my_set))
```

```python
my_empty_set=set()
print(my_empty_set)
print(type(my_empty_set))
```

```python
set([1,2,3,4,5])
print(set([1,2,3,4,5]))
print(type(set([1,2,3,4,5])))
set((1,2,3,4,5))
print(set((1,2,3,4,5)))
```

```python
set_with_duplicates={1,2,3,4,5,1,2,3}
print(set_with_duplicates)
```

```python
## Basic set operations
set_a={1,2,3,4,5}
set_b={4,5,6,7,8}

##adding an element to a set
set_a.add(6)
print(set_a)
##removing an element from a set
set_a.remove(6)
print(set_a)
# set_a.remove(10) ## this will raise a KeyError because 10 is not in the set
set_a.discard(10) ## this will not raise an error even if 10 is
```

```python
##popping an element from a set
popped_element=set_a.pop()
print(popped_element)
print(set_a)
```

```python
## clearing a set
set_a.clear()
print(set_a)
set_b.clear()
print(set_b)
```

Set Membership Test

```python
my_set={1,2,3,4,5}
print(3 in my_set)
print(6 in my_set)
```

```python
set_a={1,2,3,4,5}
set_b={4,5,6,7,8}
## union of two sets
union_set=set_a.union(set_b)
print(union_set)
## intersection of two sets
intersection_set=set_a.intersection(set_b)
print(intersection_set)
## difference of two sets
difference_set=set_a.difference(set_b)
print(difference_set)
## symmetric difference of two sets
symmetric_difference_set=set_a.symmetric_difference(set_b)
print(symmetric_difference_set)

set_a.intersection_update(set_b)
print(set_a)
```

```python
## sets methods
set_a={1,2,3,4,5}
set_b={4,5,6,7,8}

## issubset
print(set_a.issubset(set_b))
print(set_b.issubset(set_a))
## issuperset
print(set_a.issuperset(set_b))
print(set_b.issuperset(set_a))
## isdisjoint
print(set_a.isdisjoint(set_b))
print(set_a.isdisjoint({6,7,8}))
```

```python
## removing duplicates by converting a list to a set and back to a list
lst=[1,2,3,4,5,5,5,6,7,8,9,10]
my_set=set(lst)
print(my_set)
my_list=list(my_set)
print(my_list)
```

```python
## counting unique words in a text
text="hello world hello"
words=text.split()
unique_words=set(words)
print(unique_words)
print(len(unique_words))
```

Sets are a powerful and flexible data type in Python that provide a way to store collections of unique elements. They support various operations such as union, intersection, difference, and symmetric difference, which are useful for mathematical computations. Understanding how to use sets and their associated methods can help you write more efficient and clean Python code, especially when dealing with unique collections and membership tests.
