```python
df_copy=df.copy()
```

```python
df_copy.shape
```

```python
df_copy.drop(df_copy.index[10472],inplace=True)
```

```python
df_copy[~df_copy['Reviews'].str.isnumeric()]
```

```python
df_copy['Reviews']=df_copy['Reviews'].astype(int)
```

```python
df_copy.info()
```

```python
df_copy['Size'].unique()
```

```python
df_copy['Size'].isnull().sum()
```

```python
df_copy['Size']=df_copy['Size'].str.replace('M','000')
df_copy['Size']=df_copy['Size'].str.replace('k','')
df_copy['Size']=df_copy['Size'].replace('Varies with device',np.nan)
df_copy['Size']=df_copy['Size'].astype(float)
```

```python
df_copy.info()
```

```python
df_copy['Size']
```

```python
df_copy['Installs'].unique()
```

```python
df_copy['Price'].unique()
```

```python
chars_to_remove=['+','$',',']
columns_to_clean=['Installs','Price']
for item in chars_to_remove:
    for col in columns_to_clean:
        df_copy[col]=df_copy[col].str.replace(item,'')
```

```python
df_copy['Price'].unique()
```

```python
df_copy['Installs'].unique()
```

```python
df_copy['Price']=df_copy['Price'].astype(float)
df_copy['Installs']=df_copy['Installs'].astype(int)
```

```python
df_copy.info()
```

```python
## Handle last updated column
df_copy['Last Updated'].unique()
```

```python
df_copy['Last Updated']=pd.to_datetime(df_copy['Last Updated'])

df_copy['Day']=df_copy['Last Updated'].dt.day
df_copy['Month']=df_copy['Last Updated'].dt.month
df_copy['Year']=df_copy['Last Updated'].dt.year
```

```python
df_copy['Last Updated']
```

```python
df_copy.info()
```

```python
df_copy.to_csv('data/googleplaystore_cleaned.csv',index=False)
```

## Explore Data

```python
numeric_features=[feature for feature in df_copy.columns if df_copy[feature].dtype!='O']
categorical_features=[feature for feature in df_copy.columns if df_copy[feature].dtype=='O']

##print columns
print('We have {} numerical features : {}'.format(len(numeric_features),numeric_features))
print('We have {} categorical features : {}'.format(len(categorical_features),categorical_features))
```

```python
## proportion of count data on categorical features
for feature in categorical_features:
    print(df_copy[feature].value_counts(normalize=True)*100)
    print('---------------------------------------')
```

```python
plt.figure(figsize=(15,15))
plt.suptitle('Univariate analysis of numerical features',fontsize=20,fontweight='bold',alpha=0.8,y=1.0)

for i in range(len(numeric_features)):
    plt.subplot(5,3,i+1)
    sns.kdeplot(df_copy[numeric_features[i]],shade=True,color='r')
    plt.xlabel(numeric_features[i],fontsize=12)
    plt.tight_layout()
```

## Observations

Rating and Year is left skewed while Reviews,Size,Installs and Price are Right skewed

```python
plt.figure(figsize=(20,15))
plt.suptitle('Univariate analysis of categorical features',fontsize=20,fontweight='bold',alpha=0.8,y=1.0)
category=['Type','Content Rating']
for i in range(0,len(category)):
    plt.subplot(2,2,i+1)
    sns.countplot(df_copy[category[i]],palette='Set2')
    plt.xlabel(category[i])
    plt.xticks(rotation=45)
    plt.tight_layout()
```

```python
## Top 10 App Categories
category = pd.DataFrame(df_copy['Category'].value_counts())        #Dataframe of apps on the basis of category
category.rename(columns = {'Category':'Count'},inplace=True)
```
