## Lambda Functions

```python
## syntax
lambda arguments: expression
```

```python
def add(x, y):
    return x + y
print(add(2, 3))  # Output: 5
```

```python
add_lambda = lambda a, b: a + b
type(add_lambda)  # Output: <class 'function'>
print(add_lambda(2, 3))  # Output: 5
print(type(add_lambda))  # Output: <class 'function'>
```

```python
def check_even_odd(num):
    """Determines if a number is even or odd."""
    if num % 2 == 0:
        return f"{num} is even."
    else:
        return f"{num} is odd."
print(check_even_odd(10))  # Output: 10 is even.
print(check_even_odd(15))  # Output: 15 is odd.
```

```python
even_or_odd = lambda n: f"{n} is even." if n % 2 == 0 else f"{n} is odd."
print(even_or_odd(10))  # Output: 10 is even.
print(even_or_odd(15))  # Output: 15 is odd.
```

```python
def addition(x,y,z):
    return x+y+z
print(addition(1,2,3))  # Output: 6

addition_lambda = lambda x, y, z: x + y + z
print(addition_lambda(1, 2, 3))  # Output: 6
```

```python
## map() function with lambda- applies a function to all items in an iterable
numbers = [1, 2, 3, 4, 5]
def square(number):
    return number ** 2
square(2)
```

```python
list(map(lambda x: x ** 2,numbers))
```
