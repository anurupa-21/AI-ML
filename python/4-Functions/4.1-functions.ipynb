```python
## why functions are useful
# 1. Reusability: Functions allow you to reuse code, which can save time
# 2. Modularity: Functions help break down complex problems into smaller, more manageable pieces
# 3. Readability: Functions can make code easier to read and understand by giving meaningful

num=24
if num % 2 == 0:
    print(f"{num} is even.")
else:    
    print(f"{num} is odd.")
```

```python
def even_or_odd(num):
    """Determines if a number is even or odd."""
    if num % 2 == 0:
        return f"{num} is even."
    else:
        return f"{num} is odd."
```

```python
## calling the function
result = even_or_odd(24)
print(result)
print(even_or_odd(15))
```

```python
## functions with multiple parameters
def add_numbers(a, b):
    """Returns the sum of two numbers."""
    return a + b
print(add_numbers(5, 3))
```

```python
## default parameter values
def greet(name, greeting="Hello"):
    """Greets a person with a given greeting."""
    return f"{greeting}, {name}!"
print(greet("Alice"))
print(greet("Bob", "Hi"))
```

```python
## variable length arguments
#positional variable length arguments
# *args allows you to pass a variable number of positional arguments to a function.
def print_numbers(*args):
    """Prints all the numbers passed as arguments."""
    for num in args:
        print(num)

print_numbers(1, 2, 3, 4, 5)
print_numbers(10, 20)
print_numbers("Alice", "Bob", "Charlie")

# keyword variable length arguments
# **kwargs allows you to pass a variable number of keyword arguments to a function.
def print_person_info(**kwargs):
    """Prints the person's information passed as keyword arguments."""
    for key, value in kwargs.items():
        print(f"{key}: {value}")

print_person_info(name="Alice", age=30, city="New York")
print_person_info(name="Bob", age=25)
```

```python
def print_details(*args, **kwargs):
    """Prints both positional and keyword arguments."""
    print("Positional arguments:")
    for arg in args:
        print(arg)
    
    print("\nKeyword arguments:")
    for key, value in kwargs.items():
        print(f"{key}: {value}")
print_details(1, 2, 3, name="Alice", age=30)
print_details("Hello", "World", language="Python", version=3.8)
```

```python
## return statement
def calculate_area(radius):
    """Calculates the area of a circle given its radius."""
    import math
    area = math.pi * radius ** 2
    return area

# Example usage
radius = 5
area = calculate_area(radius)
print(f"The area of the circle with radius {radius} is {area:.2f}")
```

```python
def multiply(a, b):
    """Returns the product of two numbers."""
    return a * b
result = multiply(4, 5)
print(f"The product of 4 and 5 is {result}")
```
